## Disease Background:

- **Prevalence and incidence:** Facioscapulohumeral muscular dystrophy (FSHD) is a disease characterized by muscle weakness. Signs and symptoms usually appear during adolescence, but the age of onset and severity vary widely. According to the National Library of Medicine, approximately 1 in 20,000 people have FSHD; about 95% of cases are FSHD1, and the remaining 5% are FSHD2. DelveInsight estimated that 33,500 people were living with FSHD in the United States in 2021. These are prevalence estimates. A Dutch study estimated an annual incidence of approximately 3 newly diagnosed cases per million people. Sources: [MedlinePlus Genetics](https://medlineplus.gov/genetics/condition/facioscapulohumeral-muscular-dystrophy/), [DelveInsight](https://www.delveinsight.com/report-store/facioscapulohumeral-muscular-dystrophy-epidemiology-forecast), and [Deenen et al., 2014](https://pubmed.ncbi.nlm.nih.gov/25122204/).

- **Symptoms:** FSHD often affects the face, shoulders, and upper arms, but it can also affect the legs and other muscles. People may have difficulty lifting their arms, closing their eyes, smiling, walking, or climbing stairs. Weakness may be more severe on one side than the other. Weak shoulder muscles can also cause the shoulder blades to stick out. Additional symptoms may include hearing loss and retinal abnormalities that can affect vision. Source: [MedlinePlus Genetics](https://medlineplus.gov/genetics/condition/facioscapulohumeral-muscular-dystrophy/).

- **Diagnosis:** Doctors examine the pattern of muscle weakness and use genetic testing to help confirm FSHD. MRI can show changes within a muscle, but an MRI fat-fraction value alone does not diagnose the disease. Doctors consider symptoms, family history, and test results and may perform additional tests to rule out other conditions. Source: [GeneReviews: FSHD](https://www.ncbi.nlm.nih.gov/books/NBK1443/).

- **Current care:** Treatment focuses on managing symptoms and helping people maintain independence. Physical therapy, appropriate exercise, pain management, and braces for foot weakness can help with movement and daily activities. Some patients need breathing support, while selected patients may benefit from surgery to stabilize the shoulder blades. These approaches address the effects of the disease rather than its underlying genetic cause. Source: [GeneReviews: FSHD](https://www.ncbi.nlm.nih.gov/books/NBK1443/).

- **Progression and prognosis:** FSHD is a slowly progressing disease, but its severity and rate of progression vary among patients. For some, it may cause only mild muscle weakness, while others may lose the ability to walk due to FSHD. Generally, FSHD does not affect life expectancy, although complications involving breathing can occur in some cases. Since the disease is lifelong and progresses slowly, it is important to evaluate changes over an extended period using multiple measurements. Source: [MedlinePlus Medical Encyclopedia](https://medlineplus.gov/ency/article/000707.htm).


- **Biological mechanisms:** FSHD involves abnormal activity of the DUX4 gene in skeletal muscle. In FSHD1, a shortened region of repetitive DNA called D4Z4 allows DUX4 to become active when it would normally be silenced. This abnormal activity contributes to muscle damage and weakness, leading to muscle atrophy, inflammation, and the replacement of muscle tissue with fat and scar tissue. Sources: [MedlinePlus Genetics](https://medlineplus.gov/genetics/condition/facioscapulohumeral-muscular-dystrophy/)

- **Clinical trials and future treatments:** Current research is focused on addressing the underlying causes of FSHD. One example is delpacibart braxlosiran (del-brax) from Novartis, which belongs to a new class of RNA therapeutics and has shown potential to become the first disease-modifying treatment for FSHD. It has met its primary biomarker endpoint in Phase I/II clinical trials, and Phase III is currently underway. The company states, "Combining the tissue specificity of monoclonal antibodies with the precision of oligonucleotides, the AOC platform enables targeted delivery of siRNA to suppress DUX4 expression in previously hard-to-reach muscle cells of FSHD patients." Source: [Novartis clinical-trial update](https://www.novartis.com/news/media-releases/novartis-delpacibart-braxlosiran-del-brax-phase-iii-study-facioscapulohumeral-muscular-dystrophy-fshd-meets-primary-biomarker-endpoint).

## MRI Data Analysis Methdology:

### MRI Analysis Overview

The purpose of this analysis is to calculate the fat fraction of the tibialis anterior muscle using MRI water and fat images. Fat fraction provides a measurement of muscle composition by estimating the proportion of fat signal relative to the combined water and fat signal. This measurement is relevant to FSHD because progressive muscle deterioration can result in the replacement of muscle tissue with fat.

We will use Python to load the MRI images and segmentation masks provided in the dataset. The segmentation masks will allow us to isolate the tibialis anterior muscle from surrounding tissues. We will calculate fat fraction for the entire supplied image region and for the tibialis anterior muscle specifically. We will then visualize the images, examine the distribution of fat fraction values, and calculate summary statistics.

Finally, we will compare fat fraction measurements across different locations along the muscle and between the two available MRI time points, TP1 and TP2. This will allow us to investigate how muscle composition varies by location and visit.

### Loading the MRI Dataset

We begin by importing the Python libraries needed for our analysis. NumPy will be used to process the MRI images, Pandas will organize the information associated with each MRI slice, and Matplotlib will be used to visualize the results. The slice information file identifies the subject, time point, sampling location, and corresponding water, fat, and segmentation mask files.

```python
from pathlib import Path
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

# Locate the MRI data folder
DATA_DIR = Path("FSHD_MRI_data/FSHD_TA_MRI_data_0020_TP1_TP2")

# Load the MRI slice information
manifest = pd.read_csv(DATA_DIR / "slice_info.csv")

# Display the dataset information
display(manifest)
```

### Loading MRI Images and Segmentation Masks

We begin by examining the first MRI slice from TP1. Each slice contains a water image, a fat image, and a segmentation mask identifying the tibialis anterior muscle. The water and fat images contain MRI signal intensities, while the segmentation mask identifies which pixels belong to the muscle. By applying the mask, we can calculate fat fraction specifically within the tibialis anterior rather than including surrounding tissues.

```python
# Select the first MRI slice from TP1
slice_data = manifest[manifest["timepoint"] == "TP1"].iloc[0]

# Load the water, fat, and segmentation mask images
water = np.load(DATA_DIR / slice_data["water_file"])
fat = np.load(DATA_DIR / slice_data["fat_file"])
mask = np.load(DATA_DIR / slice_data["mask_file"])

# Convert the segmentation mask to Boolean values
muscle_mask = mask > 0

# Display image information
print("Water image shape:", water.shape)
print("Fat image shape:", fat.shape)
print("Mask shape:", mask.shape)
print("Number of muscle pixels:", np.sum(muscle_mask))
```

### Calculating Fat Fraction

Fat fraction is calculated by dividing the fat signal intensity by the combined fat and water signal intensities. This calculation produces a percentage representing the proportion of the total signal associated with fat. To avoid division by zero, pixels where the combined water and fat signal is zero are excluded from the calculation. The segmentation mask is then used to isolate fat fraction values within the tibialis anterior muscle.

```python
# Calculate the combined water and fat signal
total_signal = water + fat

# Calculate fat fraction while avoiding division by zero
fat_fraction = np.divide(
    fat,
    total_signal,
    out=np.zeros_like(fat, dtype=float),
    where=total_signal > 0
) * 100

# Extract fat fraction values within the tibialis anterior
muscle_fat_fraction = fat_fraction[muscle_mask]

# Calculate summary statistics
mean_ff = np.mean(muscle_fat_fraction)
median_ff = np.median(muscle_fat_fraction)
std_ff = np.std(muscle_fat_fraction)

print(f"Mean fat fraction: {mean_ff:.2f}%")
print(f"Median fat fraction: {median_ff:.2f}%")
print(f"Standard deviation: {std_ff:.2f}%")
print(f"Number of muscle pixels: {len(muscle_fat_fraction)}")
```

### MRI Visualization

To visualize the MRI data, we display the water image, fat image, segmentation mask, and calculated fat fraction map. The water and fat images show the distribution of their respective MRI signals, while the segmentation mask identifies the tibialis anterior muscle. The fat fraction map displays the estimated percentage of fat signal at each pixel. Comparing these images helps demonstrate how the original MRI data are used to calculate muscle fat fraction.

```python
# Create a figure with four MRI images
fig, axes = plt.subplots(1, 4, figsize=(16, 4))

# Water image
axes[0].imshow(water, cmap="gray")
axes[0].set_title("Water Image")

# Fat image
axes[1].imshow(fat, cmap="gray")
axes[1].set_title("Fat Image")

# Segmentation mask
axes[2].imshow(muscle_mask, cmap="gray")
axes[2].set_title("Tibialis Anterior Mask")

# Fat fraction map
im = axes[3].imshow(fat_fraction, cmap="viridis", vmin=0, vmax=100)
axes[3].set_title("Fat Fraction Map")

# Remove axes for clearer MRI visualization
for ax in axes:
    ax.axis("off")

# Add a color bar showing fat fraction percentage
fig.colorbar(im, ax=axes[3], label="Fat Fraction (%)")

plt.suptitle("TP1: Tibialis Anterior MRI Analysis")
plt.show()
```

### Interpretation of Initial MRI Results

The first MRI slice from TP1 produced a mean fat fraction of 83.92% and a median fat fraction of 88.83% within the segmented tibialis anterior muscle. The standard deviation was 13.27 percentage points, indicating variability in fat fraction across the muscle pixels. The segmentation mask contained 61 pixels, allowing the analysis to focus specifically on the tibialis anterior rather than surrounding tissues.

The relatively high fat fraction suggests that the segmented region contains a substantial fat signal compared with its water signal. However, these results represent only one slice from one time point. Additional slices and the second time point must be analyzed to better understand how fat fraction varies along the muscle and between visits.

### Distribution of Muscle Fat Fraction

To further examine the variation in fat fraction, we created a histogram of the fat fraction values within the segmented tibialis anterior muscle. This allows us to visualize the frequency of different fat fraction percentages and determine whether the values are concentrated around the mean or distributed across a wider range.

```python
# Plot the distribution of fat fraction within the muscle
plt.figure(figsize=(8, 5))

plt.hist(muscle_fat_fraction, bins=15, edgecolor="black")

# Add lines for the mean and median
plt.axvline(mean_ff, linestyle="--", label=f"Mean: {mean_ff:.2f}%")
plt.axvline(median_ff, linestyle=":", label=f"Median: {median_ff:.2f}%")

plt.xlabel("Fat Fraction (%)")
plt.ylabel("Number of Muscle Pixels")
plt.title("Distribution of Tibialis Anterior Fat Fraction (TP1, Slice 01)")
plt.legend()

plt.show()
```

### Interpretation of Fat Fraction Distribution

The histogram shows that most pixels within the segmented tibialis anterior muscle have fat fraction values between approximately 80% and 100%. The mean fat fraction was 83.92%, while the median was 88.83%. The distribution is left-skewed, with several pixels having lower fat fractions that decrease the mean. This suggests that the fat signal is relatively high throughout much of the segmented region, although there is some variation between pixels. These results describe only the first slice at TP1 and cannot establish whether fat fraction changes over time.

### Fat Fraction Across MRI Slices and Time Points

To investigate how fat fraction varies along the tibialis anterior muscle, we calculated the mean fat fraction for each of the eight MRI slices at TP1 and TP2. Each slice represents a different location along the muscle. The segmentation masks were used to isolate the tibialis anterior, and the mean fat fraction was calculated using only pixels within the segmented region. The resulting measurements allow us to compare muscle composition across locations and between the two MRI visits.

```python
# Store the results for each MRI slice
results = []

# Analyze every slice in the dataset
for _, row in manifest.iterrows():

    # Load the MRI images and segmentation mask
    water = np.load(DATA_DIR / row["water_file"])
    fat = np.load(DATA_DIR / row["fat_file"])
    mask = np.load(DATA_DIR / row["mask_file"])

    # Identify pixels belonging to the tibialis anterior
    muscle_mask = mask > 0

    # Calculate fat fraction
    total_signal = water + fat

    fat_fraction = np.divide(
        fat,
        total_signal,
        out=np.zeros_like(fat, dtype=float),
        where=total_signal > 0
    ) * 100

    # Extract fat fraction values within the muscle
    muscle_values = fat_fraction[muscle_mask]

    # Calculate summary statistics
    results.append({
        "Time Point": row["timepoint"],
        "Muscle Location (%)": row["percent_along_muscle"],
        "Mean Fat Fraction (%)": np.mean(muscle_values),
        "Median Fat Fraction (%)": np.median(muscle_values),
        "Standard Deviation": np.std(muscle_values),
        "Muscle Pixels": len(muscle_values)
    })

# Convert results into a DataFrame
results_df = pd.DataFrame(results)

# Display the results
display(results_df.round(2))
```

```python
# Verify segmentation masks for all 16 MRI slices
for _, row in manifest.iterrows():
    mask_check = np.load(DATA_DIR / row["mask_file"])
    labels = np.unique(mask_check)
    print(row["timepoint"], row["percent_along_muscle"], labels)
```

### Comparison of Fat Fraction Across Muscle Locations

To compare muscle composition between the two MRI visits, we plotted the mean fat fraction at each sampled location along the tibialis anterior muscle. This allows us to examine whether fat fraction varies along the muscle and whether differences are observed between TP1 and TP2.

```python
# Separate results by time point
tp1 = results_df[results_df["Time Point"] == "TP1"]
tp2 = results_df[results_df["Time Point"] == "TP2"]

# Sort measurements by muscle location
tp1 = tp1.sort_values("Muscle Location (%)")
tp2 = tp2.sort_values("Muscle Location (%)")

# Plot mean fat fraction for both visits
plt.figure(figsize=(9, 5))

plt.plot(
    tp1["Muscle Location (%)"],
    tp1["Mean Fat Fraction (%)"],
    marker="o",
    label="TP1"
)

plt.plot(
    tp2["Muscle Location (%)"],
    tp2["Mean Fat Fraction (%)"],
    marker="s",
    label="TP2"
)

plt.xlabel("Location Along Tibialis Anterior (%)")
plt.ylabel("Mean Fat Fraction (%)")
plt.title("Tibialis Anterior Fat Fraction Across Muscle Locations")
plt.legend()
plt.grid(alpha=0.3)

plt.show()
```

### Interpretation of Fat Fraction Across Muscle Locations

The graph shows that mean fat fraction varies considerably along the tibialis anterior muscle. At both time points, fat fraction was highest near the beginning of the sampled muscle length and generally decreased toward 60% to 70% of muscle length. Fat fraction increased again at the 80% location.

TP2 generally showed higher mean fat fraction values than TP1, particularly between 30% and 60% of muscle length. For example, at 30% of muscle length, mean fat fraction increased from approximately 58% at TP1 to 82% at TP2. This suggests a difference in muscle composition between the two visits.

However, these measurements represent a single participant, and differences between visits may also be influenced by image acquisition, segmentation, or sampling variability. Therefore, additional measurements and participants would be needed to determine whether these changes reliably reflect FSHD progression.

### Overall Comparison Between MRI Visits

To summarize the differences between MRI visits, we calculated the average of the eight mean fat fraction measurements for TP1 and TP2. We then calculated the difference between the two time points. This provides a summary of the average fat fraction across the sampled muscle locations, allowing us to quantify the overall difference between visits.

```python
# Calculate the average of the eight slice means
tp1_mean = tp1["Mean Fat Fraction (%)"].mean()
tp2_mean = tp2["Mean Fat Fraction (%)"].mean()

# Calculate the change between visits
difference = tp2_mean - tp1_mean

# Display results
print(f"TP1 average fat fraction: {tp1_mean:.2f}%")
print(f"TP2 average fat fraction: {tp2_mean:.2f}%")
print(f"Difference between visits: {difference:+.2f} percentage points")
```

### Interpretation of Overall MRI Results

The average fat fraction across the eight sampled muscle locations increased from 42.26% at TP1 to 54.41% at TP2, representing an increase of 12.15 percentage points. This suggests that the tibialis anterior muscle had a greater proportion of fat signal at the second MRI visit.

The results also showed that fat fraction varied considerably along the length of the muscle, indicating that muscle composition was not uniform across the sampled locations. These findings suggest that MRI fat fraction may be a useful measurement for evaluating changes in muscle composition in patients with FSHD.

However, this analysis only examined one participant, and the averages were calculated by giving equal weight to each of the eight sampled slices. Therefore, these results should not be interpreted as representing the entire muscle volume or the broader FSHD population. Additional participants and measurements would be necessary to determine whether the observed increase consistently reflects disease progression.

## Interpolation Analysis:

### Interpolation of Tibialis Anterior Fat Fraction

The MRI dataset contains fat fraction measurements at eight locations along the tibialis anterior muscle, ranging from 10% to 80% of muscle length in increments of 10%. However, fat fraction values are not directly measured at locations between these sampled points.

To estimate fat fraction at an unsampled location, we will use linear interpolation. This method estimates a value between two known measurements by assuming that fat fraction changes at a constant rate between those locations.

For this analysis, we will estimate the mean fat fraction at 45% of muscle length using the measured values at 40% and 50%. We will perform this calculation for both TP1 and TP2 to compare the estimated fat fraction between visits.

Because 45% falls between two measured locations, this calculation is interpolation rather than extrapolation.

```python
# Define the location where fat fraction will be estimated
target_location = 45

# Interpolate fat fraction at 45% muscle length for TP1
tp1_interpolated = np.interp(
    target_location,
    tp1["Muscle Location (%)"],
    tp1["Mean Fat Fraction (%)"]
)

# Interpolate fat fraction at 45% muscle length for TP2
tp2_interpolated = np.interp(
    target_location,
    tp2["Muscle Location (%)"],
    tp2["Mean Fat Fraction (%)"]
)

# Display the estimated fat fractions
print(f"TP1 estimated fat fraction at 45%: {tp1_interpolated:.2f}%")
print(f"TP2 estimated fat fraction at 45%: {tp2_interpolated:.2f}%")
```

### Visualization of Interpolated Fat Fraction

To visualize the interpolation results, we plotted the measured mean fat fraction values across the eight sampled locations for both MRI visits. We then added the interpolated estimates at 45% of muscle length.

The measured values are connected using straight lines to represent linear interpolation between sampled locations. The interpolated estimates are highlighted separately to distinguish calculated estimates from actual MRI measurements.

```python
# Create a figure for the interpolation analysis
plt.figure(figsize=(9, 5))

# Plot measured fat fraction values for TP1
plt.plot(
    tp1["Muscle Location (%)"],
    tp1["Mean Fat Fraction (%)"],
    marker="o",
    label="TP1 Measured"
)

# Plot measured fat fraction values for TP2
plt.plot(
    tp2["Muscle Location (%)"],
    tp2["Mean Fat Fraction (%)"],
    marker="s",
    label="TP2 Measured"
)

# Highlight the interpolated estimates at 45%
plt.scatter(
    target_location,
    tp1_interpolated,
    marker="X",
    s=150,
    label=f"TP1 Estimated: {tp1_interpolated:.2f}%"
)

plt.scatter(
    target_location,
    tp2_interpolated,
    marker="X",
    s=150,
    label=f"TP2 Estimated: {tp2_interpolated:.2f}%"
)

# Add a vertical reference line at the estimated location
plt.axvline(
    target_location,
    linestyle="--",
    alpha=0.5
)

# Label the graph
plt.xlabel("Location Along Tibialis Anterior (%)")
plt.ylabel("Mean Fat Fraction (%)")
plt.title("Linear Interpolation of Tibialis Anterior Fat Fraction")

plt.legend()
plt.grid(alpha=0.3)

plt.show()
```

### Interpretation of Interpolation Results

Linear interpolation was used to estimate the mean fat fraction at 45% of the tibialis anterior muscle length. The estimated fat fraction was 25.40% at TP1 and 46.30% at TP2, representing a difference of 20.90 percentage points between the two visits.

The estimates were calculated using the measured mean fat fraction values at 40% and 50% of muscle length. Because 45% is halfway between these sampled locations, linear interpolation estimates the fat fraction by taking the average of the two neighboring measurements.

The interpolation method assumes that fat fraction changes at a constant rate between the sampled locations. However, the actual distribution of fat within the muscle may not follow a straight line. Therefore, the interpolated values are estimates rather than new MRI measurements.

Both estimates fall within the range of the sampled locations, meaning interpolation was used rather than extrapolation. Extrapolation would involve estimating values outside the measured range, where predictions would generally be more uncertain.

Several factors could affect the accuracy of these estimates, including differences in MRI acquisition, segmentation accuracy, and variation in muscle composition between sampled locations. Because no MRI measurement was collected at exactly 45% of muscle length, we cannot determine the accuracy of the interpolated estimates without additional measurements.

The higher estimated fat fraction at TP2 is consistent with the overall increase observed between visits. However, additional measurements would be needed to determine whether this difference reliably reflects FSHD progression.
